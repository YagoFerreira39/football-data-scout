```python
def safe_divide(numerator: pd.Series, denominator: pd.Series) -> pd.Series:
    denominator = denominator.replace(0, np.nan)
    return numerator / denominator


def add_wyscout_derived_columns(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()

    # xA per 100 passes
    # Uses per 90 values, which is valid because the per-90 denominator cancels out.
    df["xA per 100 passes"] = (
        safe_divide(df["xA per 90"], df["Passes per 90"]) * 100
    )

    # xA per cross
    # Measures creative output relative to crossing volume.
    df["xA per Cross"] = safe_divide(
        df["xA per 90"],
        df["Crosses per 90"],
    )

    # % of passes that are crosses
    df["Crosses share of passes, %"] = (
        safe_divide(df["Crosses per 90"], df["Passes per 90"]) * 100
    )

    # Key passes per 100 passes
    df["Key passes per 100 passes"] = (
        df["Key passes per 90"] / df["Passes per 90"].replace(0, np.nan)
    ) * 100

    # Non-Penalty xG
    # Approximation only, because the dataset has Penalties taken
    # but does not provide Penalty xG directly.
    #
    # Common penalty xG approximation: 0.76 per penalty.
    df["Non-Penalty xG"] = df["xG"] - (df["Penalties taken"] * 0.76)

    # Prevent negative values in edge cases
    df["Non-Penalty xG"] = df["Non-Penalty xG"].clip(lower=0)

    df["Non-Penalty xG per 90"] = (
        safe_divide(df["Non-Penalty xG"], df["Minutes played"]) * 90
    )

    return df

# df = add_wyscout_derived_columns(df=df)
# filtered_df = df[df["League"] == "England Premier League 2025-26"].copy()
```

```python
# Filter by position
# Below are the positions that Wyscout have.
# Fullback or Wingback use RB, LB, RWB, LWB
# Centre Back use CB
# Midfielder use DMF, CMF
# Attacking midfielder use AMF
# Wingers use RW, LW, RAMF, LAMF, WF
# Strikers use CF

# Define the positions to filter
cb_positions = ['CB']
fb_wb_positions = ['RB', 'LB', 'RWB', 'LWB']
mf_positions = ['DMF', 'CMF']
am_positions = ['AMF']
w_positions = ['RW', 'LW', 'RAMF', 'LAMF', 'WF']
st_positions = ['CF']
minimum_minutes_played = 1000

def get_columns_to_zscore(df: pd.DataFrame) -> list[str]:
    # Columns to exclude from z-score conversion
    exclude_columns = [
        'Unnamed: 0', 'Player', 'Team', 'Team within selected timeframe', 'Position',
        'Age', 'Market value', 'Contract expires', 'Matches played', 'Minutes played',
        'Birth country', 'Passport country', 'Foot', 'Height', 'Weight', 'On loan',
        'League', 'Main Position'
    ]

    # Columns to apply z-score conversion
    columns_to_zscore = [col for col in df.columns if col not in exclude_columns]

    return columns_to_zscore

def prepare_df(wyscout_df: pd.DataFrame, league: str, season: str, positions: list[str]) -> pd.DataFrame:
    df = wyscout_df.copy()
    df = add_wyscout_derived_columns(df=df)
    filter_league = f"{league} {season}"
    filtered_df = df[df["League"] == filter_league].copy()


    # Filter the DataFrame based on the "Main Position" column
    filtered_df = filtered_df[filtered_df['Main Position'].isin(positions)]

    # Filtering for minimum number of minutes played
    filtered_df = filtered_df[filtered_df['Minutes played'] > minimum_minutes_played]

    # # Columns to exclude from z-score conversion
    # exclude_columns = [
    #     'Unnamed: 0', 'Player', 'Team', 'Team within selected timeframe', 'Position',
    #     'Age', 'Market value', 'Contract expires', 'Matches played', 'Minutes played',
    #     'Birth country', 'Passport country', 'Foot', 'Height', 'Weight', 'On loan',
    #     'League', 'Main Position'
    # ]

    # # Columns to apply z-score conversion
    # columns_to_zscore = [col for col in filtered_df.columns if col not in exclude_columns]
    columns_to_zscore = get_columns_to_zscore(df=filtered_df)

    # Applying z-score normalisation to the selected columns
    filtered_df[columns_to_zscore] = filtered_df[columns_to_zscore].apply(zscore)

    return filtered_df
```

```python
# # Function to normalize the data to a [0, 1] range
# def normalize_data(data):
#     # Subtract the minimum value from each element and divide by the range
#     return (data - np.min(data)) / (np.max(data) - np.min(data))
def normalize_data(series: pd.Series) -> pd.Series:
    min_value = series.min()
    max_value = series.max()

    if max_value == min_value:
        return pd.Series(0.0, index=series.index)

    return (series - min_value) / (max_value - min_value)

def normalize_zscore_columns(df: pd.DataFrame) -> pd.DataFrame:
    # Ensure you're working with a copy of the original DataFrame
    # This avoids accidental changes to the original data and avoids the SettingWithCopyWarning
    df = df.copy()
    columns_to_zscore = get_columns_to_zscore(df=df)

    # Iterate over the specified columns (columns_to_zscore) to normalize them
    for column in columns_to_zscore:
        # Step 1: Shift the values in the column to ensure no negative values
        # Add the absolute value of the column's minimum to all values in the column
        # This ensures the minimum value becomes 0
        df[column] = df[column] + abs(df[column].min())
        
        # Step 2: Normalize the column to a [0, 1] range
        # Apply the NormalizeData function to scale the column values
        df[column] = normalize_data(series=df[column])

    return df

def create_role_rating_columns(
    df: pd.DataFrame,
    feature_weights: dict,
) -> pd.DataFrame:
    """
    Creates weighted metric columns, category rating columns,
    and optionally a final role rating column.
    """
    new_columns = {}

    for category, metrics in feature_weights.items():
        category_att_score = 0.0
        for key, value in metrics.items():
            column_name = f"{key} Rating"
            new_columns[column_name] = df[key] * value
            category_att_score += new_columns[column_name]
        column_name = f"{category} Att"
        new_columns[column_name] = category_att_score

    df = pd.concat([df, pd.DataFrame(new_columns)], axis=1)

    return df

fixed_columns = [
        "Player",
        "Team within selected timeframe",
        "Main Position",
        "Age",
        "Minutes played",
    ]
# att_columns = [col for col in filtered_df.columns if col.endswith("Att")]

# def create_ratings_df(filtered_df: pd.DataFrame, fixed_columns: list[str], att_columns: list[str]) -> pd.DataFrame:
    
#     selected_columns = fixed_columns + att_columns

#     ratings_df = filtered_df[selected_columns].copy()

#     return ratings_df

def create_ratings_df(
    df: pd.DataFrame,
    fixed_columns: list[str],
) -> pd.DataFrame:
    att_columns = [col for col in df.columns if col.endswith("Att")]

    selected_columns = fixed_columns + att_columns

    return df[selected_columns].copy()

def normalize_attribute_columns(ratings_df: pd.DataFrame) -> pd.DataFrame:

    # Ensure a copy of the DataFrame to prevent unintended changes to the original data
    df = ratings_df.copy()

    # Get the column names from the 5th column onwards
    att_columns = [col for col in df.columns if col.endswith("Att")]
    columns_to_normalize = att_columns  # Index starts at 0, so 5 corresponds to the 6th column

    # Iterate over the selected columns and apply normalization
    for column in columns_to_normalize:
        # Step 1: Shift values to ensure no negative numbers
        # Add the absolute minimum of the column to all its values
        df[column] = df[column] + abs(df[column].min())
        
        # Step 2: Normalize the column to a [0, 1] range
        # Apply the NormalizeData function to scale the column
        df[column] = normalize_data(series=df[column])

    return df

def normalize_overall_rating(ratings_df: pd.DataFrame) -> pd.DataFrame:
    # Ensure a copy of the DataFrame to prevent unintended changes
    df = ratings_df.copy()

    # Select the last column in the DataFrame
    last_column = ratings_df.columns[-1]

    # Apply normalization to the last column
    # Step 1: Shift values to ensure no negative numbers
    ratings_df[last_column] = df[last_column] + abs(df[last_column].min())

    # Step 2: Normalize the column to a [0, 1] range
    df[last_column] = normalize_data(series=df[last_column])

    return df

def create_overall_rating(ratings_df: pd.DataFrame, category_weights: dict[str, float]) -> pd.DataFrame:
    df = ratings_df.copy()

    overall_rating = 0.0
    for category, weight in category_weights.items():
        column_att = f"{category} Att"
        df[column_att] = df[column_att] * weight
        overall_rating += df[column_att]

    df["Overall Rating"] = overall_rating

    # Ensure a copy of the DataFrame to prevent unintended changes
    ratings = df.copy()

    # Select the last column in the DataFrame
    last_column = df.columns[-1]

    # Apply normalization to the last column
    # Step 1: Shift values to ensure no negative numbers
    df[last_column] = df[last_column] + abs(df[last_column].min())

    # Step 2: Normalize the column to a [0, 1] range
    df[last_column] = normalize_data(df[last_column])

    return df

def prepare_ratings_table(ratings_df: pd.DataFrame) -> pd.DataFrame:
    # Make a copy of the ratings DataFrame to avoid accidental modifications
    df = ratings_df.copy()

    # Multiply all columns from the 6th column onwards by 100
    df.iloc[:, 5:] = df.iloc[:, 5:] * 100  # Scale 6th column onward by 100

    # Round the same columns to 1 decimal place
    df.iloc[:, 5:] = df.iloc[:, 5:].round(2)  # Round to 1 decimal place

    # Sort the DataFrame by the last column in descending order
    df.sort_values(by=df.columns[-1], ascending=False, inplace=True)  # Sort by last column

    # Reset the index of the sorted DataFrame to ensure a clean, sequential index
    df.reset_index(drop=True, inplace=True)

    return df
```

```python
filtered_df = prepare_df(wyscout_df=wyscout_df, league="England Premier League", season="2025-26", positions=st_positions)
filtered_df = normalize_zscore_columns(df=filtered_df)
filtered_df = create_role_rating_columns(
    df=filtered_df,
    feature_weights=CF_TARGET_MAN_FEATURE_WEIGHTS,
)

ratings_df = create_ratings_df(df=filtered_df, fixed_columns=fixed_columns)
ratings_df = normalize_attribute_columns(ratings_df=ratings_df)
ratings_df = normalize_overall_rating(ratings_df=ratings_df)
ratings_df = create_overall_rating(ratings_df=ratings_df, category_weights=CF_TARGET_MAN_CATEGORY_WEIGHTS)
ratings_df = prepare_ratings_table(ratings_df=ratings_df)

ratings_df
```
